## Análisis de Correlación:

### Correlación entre variables numéricas

```python
%%R

num_vars <- dfMedidas[, sapply(dfMedidas, is.numeric)]
num_vars <- num_vars[, !names(num_vars) %in% c("Rented_Bike_Count")]
corr_test <- corr.test(num_vars)

# Correlaciones
print(round(corr_test$r, 2))
```

Y de una forma más visual:

```python
%%R
num_vars <- dfMedidas[, sapply(dfMedidas, is.numeric)]
ggpairs(num_vars, size = 3)
```

```python
%%R
num_vars <- dfMedidas[, sapply(dfMedidas, is.numeric)]
cor_matrix <- cor(num_vars, use = "complete.obs")

corrplot(corr_test$r,
        type = "upper", 
        method = "color", 
        tl.col = "black", 
        tl.srt = 45,
        size = 10)
```

Podemos ver que las variables más relacionadas son la temperatura con la temperatura del punto de rocío (91%), humedad con visibilidad (-54%) y humedad con la temperatura del punto de rocío (54%). 

Estas relaciones tienen sentido, ya que, la temperatura del punto de rocío depende directamente con la temperatura ambiental, por otro lado, una mayor humedad puede derivar en una menor visibilidad por factores como la condensación (niebla, bruma, bochorno, ...), por último, la temperatura del punto de rocío se relaciona tanto con la humedad debido a que la humedad mide el vapor de agua ambiental, mientras que el punto de rocío mide la temperatura a partir de la cual esa humedad ambiental alcance el 100%.

Como se alcanza más de un 90% de correlación entre 2 variables, se ha decidido agrupar dichas variables en una PCA.

```python
%%R
vars <- dfMedidas[, c("Temperature", "Dew_point_temperature")]
vars_scaled <- scale(vars)

pca <- prcomp(vars_scaled, center = TRUE, scale. = TRUE)
print(summary(pca))

dfMedidas$temperature_index <- pca$x[, 1]
dfMedidas <- dfMedidas[,!names(dfMedidas) %in% c("Temperature", "Dew_point_temperature")]
```

Podemos ver que esta nueva variable explica el 95.6% de la varianza de las variables originales

### Correlación con la variable objetivo:

Evaluamos la correlación de las variables numéricas con la variable objetivo

```python
%%R
library(ggplot2)

target <- "Rented_Bike_Count"

num_vars <- dfMedidas[, sapply(dfMedidas, is.numeric)]
correlations <- cor(num_vars, use = "complete.obs", method = "pearson")[, target]

cor_df <- data.frame(
  Variable = names(correlations),
  Correlation = correlations
)
cor_df <- cor_df[cor_df$Variable != target, ]
cor_df <- cor_df[order(-abs(cor_df$Correlation)), ]

print(cor_df)
```

Como podemos ver la variable más relacionada con la objetivo es el índice de la temperatura (47% de correlación), es decir, cuanto mayor es la temperatura mayor es el uso de las biciletas, esto responde al comportamiento promedio del usuario de bicicletas.
Por otro lado, el resto de valores no son significativos.

## Reducción de Dimensionalidad:

Aunque ya hemos realizado un estudio sobre la correlación de los atributos, ahora analizaremos más detalladamente si se puede o no considerar reducir la dimensionalidad del problema fusionando algunos más.

```python
%%R
atributos <- c("temperature_index", "Hour", "Solar_Radiation", "Humidity", 
               "Visibility", "Snowfall", "Rainfall", "Wind_speed")

res.pca <- prcomp(dfMedidas[,atributos], scale = TRUE)
summary(res.pca)
```

# Modelos

## Regresión Lineal

```python
%%R
fit_lm <- lm(Rented_Bike_Count ~ ., data = dfMedidas)
summary(fit_lm)
```

Como podemos ver, este modelo explica el 50% del error a la hora de predecir el número de bicicletas alquiladas.

Además, podemos ver que las variables más significativas son "Functioning Day", que explica que si el día es laborable, hay una mayor demanda, y "SeasonsWinter", que explica que en invierno la demanda es menor. Además, la información meteorológica que contienen las PCA es importante, más en las "PC1", "PC2" y "PC5"
Las menos importantes son "SeasonsSummer", "PC3" y "PC4".

```python
%%R
plot(fit_lm)
```

De estos gráficos podemos sacar las siguientes conclusiones:
* Residuals vs Fitted: Clara curvatura --> Heterocedasticidad + No linealidad (El modelo no captura bien la relación)
* Q - Q: Desviaciones en las colas --> La inferencia estricta queda descartada.
* Scale - Location: La varianza de los residuos crece con el ajuste --> Heterocedasticidad.
* Residuals vs Leverage: Algunos puntos importantes --> Mejorable

Dados los malos resultados probaremos regresión no lineal.

## Regresión no lineal

### Regresión Cuadrática

```python
%%R
fit_quad <- lm(
  Rented_Bike_Count ~ 
    PC1 + I(PC1^2) +
    PC2 + I(PC2^2) +
    PC3 + I(PC3^2) + 
    PC4 + I(PC4^2) +
    PC5 + (PC5^2) +
    Holiday + Functioning_Day + SeasonsWinter + SeasonsSpring + SeasonsSummer + SeasonsAutumn,
  data = dfMedidas
)

summary(fit_quad)
```

Podemos ver que el resultado no ha mejorado mucho, habiendo mejorado la explicación del error en un 2% (50% -> 52%).

Además, podemos ver que las variables más significativas siguen siendo "Functioning Day" y "SeasonsWinter", con el mismo orden de antes en cuanto al resto de variables y su implicación en el modelo.

```python
%%R
plot(fit_quad)
```

En cuanto a los gráficos, no hay mucha diferencia. 
La diferencia más significativa estña en el gráfico "Residuals vs Leverage": Podemos ver como, al añadir los cuadrados, el modelo se ha vuelto más sensible a los términos extremos.

### Regresión Exponencial

```python
%%R
fit_exp <- lm(
  log1p(Rented_Bike_Count) ~ .,
  data = dfMedidas
)

summary(fit_exp)
```

Como podemos ver, el modelo ha mejorado mucho con la regresión exponencial, pues ahora explica el 77% del error.

En cuanto a las variables más significativas, podemos ver lo mismo que hemos visto antes, pero con valores menores debido al uso de logaritmos para estabilizar la varianza.

```python
%%R
plot(fit_exp)
```

De estos gráficos podemos sacar las siguientes conclusiones:
* Residuals vs Fitted: Nube de puntos --> Ya no se ve tanta estructura como antes, pero es mejorable.
* Q - Q: Desviaciones en las colas --> Residuos no normales y puntos extremos.
* Scale - Location: Baja heterocedasticidad, pero sigue habiendo un poco.
* Residuals vs Leverage: Algunos puntos importantes --> Mejorable


### GAM

```python
%%R
fit_gam <- gam(
  log1p(Rented_Bike_Count) ~
    s(PC1) + s(PC2) + s(PC3) + s(PC4) + s(PC5) +
    Holiday +Functioning_Day + SeasonsWinter + SeasonsSpring + SeasonsSummer + SeasonsAutumn,
  data = dfMedidas,
  method = "REML"
)

summary(fit_gam)
```

Podemos ver que el modelo GAM explica mucho mejor la varianza del error.

Por otro lado, en cuanto a laas variables importantes, vuelve a ser lo mismo de antes.

```python
%%R
par(mfrow=c(2,2))
gam.check(fit_gam)
plot(fit_gam, pages = 1)
```

De estos gráficos podemos sacar las siguientes conclusiones:
* Residuals vs Predictor: Dos estructuras --> Heterocedasticidad residual.
* Q - Q: Desviaciones en las colas --> Hay algunos valores extremos, pero en general hay una normalidad razonable.
* Scale - Location: La varianza de los residuos crece con el ajuste --> Heterocedasticidad.
* Response vs Fitted: Dispersión en valores altos --> El modelo capta bien la tendencia global.


### Random Forest

```python
%%R
rf <- ranger(
  Rented_Bike_Count ~ .,
  data = dfMedidas,
  num.trees = 500,
  importance = "permutation"
)
```

```python
%%R
pred_rf <- predict(rf, data = dfMedidas)$predictions
res_rf <- dfMedidas$Rented_Bike_Count - pred_rf
```

RMSE:

```python
%%R
rmse(dfMedidas$Rented_Bike_Count, pred_rf)
```

Podemos ver que el modelo mejora el modelo lineal (456.5 --> 195.7)

R^2:

```python
%%R
y <- dfMedidas$Rented_Bike_Count
y_hat <- pred_rf

R2_rf <- 1 - sum((y - y_hat)^2) / sum((y - mean(y))^2)
R2_rf
```

Ahora si podemos ver una mejora significativa y un modelo bueno, que explica el 91% del error.

Observado vs Predicho:

```python
%%R
plot(
  dfMedidas$Rented_Bike_Count,
  pred_rf,
  pch = 16,
  col = rgb(0,0,0,0.2),
  xlab = "Observado",
  ylab = "Predicho",
  main = "Random Forest: Observado vs Predicho"
)
abline(0,1,col="red",lwd=2)
```

Podemos ver que el modelo ajusta muy bien la estructura no lineal, pero suaviza los extremos.

Residual vs Fitted:

```python
%%R
plot(
  pred_rf, res_rf,
  pch = 16, col = rgb(0,0,0,0.1),
  xlab = "Valores ajustados (RF)",
  ylab = "Residuos",
  main = "Random Forest: Residuos vs Ajustados"
)
abline(h = 0, col = "red", lwd = 2)
```

Podemos volver a ver que para valores pequeños el modelo funciona bien, sin embargo, para valores más grandes el modelo presenta heterocedasticidad

Q - Q:

```python
%%R
qqnorm(res_rf, pch = 16, col = rgb(0,0,0,0.1))
qqline(res_rf, col = "red", lwd = 2)
```

Con este gráfico volvemos a ver que para los extremos no funciona todo lo bien que podría.
